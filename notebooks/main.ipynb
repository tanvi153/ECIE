```python
import os
from pathlib import Path
import pandas as pd

# =========================================================
# 1. DYNAMIC FILE PATH RESOLUTION
# =========================================================
# Checks project data/raw directory, env variable, or legacy path
candidate_dirs = [
    Path.cwd().parent / 'data' / 'raw',
    Path.cwd() / 'data' / 'raw',
    Path(os.getenv('DATA_DIR', os.getenv('ECIE_DATA_DIR', ''))),
    Path.cwd().parent / 'ecommerce_csvs',
    Path.cwd() / 'ecommerce_csvs',
]

data_dir = None
for d in candidate_dirs:
    if d and d.exists() and (d / 'table1_reviews.csv').exists():
        data_dir = d
        break

if data_dir is None:
    # Default to project data/raw
    data_dir = Path.cwd().parent / 'data' / 'raw' if (Path.cwd().parent / 'data').exists() else Path.cwd() / 'data' / 'raw'

print(f'Using Data Directory: {data_dir}')

# =========================================================
# 2. READ ALL 5 CSV FILES
# =========================================================
table1_reviews = pd.read_csv(data_dir / 'table1_reviews.csv')
table2_review_text = pd.read_csv(data_dir / 'table2_review_text.csv')
table3_products = pd.read_csv(data_dir / 'table3_products.csv')
table4_stores = pd.read_csv(data_dir / 'table4_stores.csv')
table5_users = pd.read_csv(data_dir / 'table5_users.csv')

print('All 5 tables loaded successfully!')
```

```python
# =========================================================
# 3. STORE ALL TABLES IN A DICTIONARY
# =========================================================

tables = {
    "Table 1 - Reviews": table1_reviews,
    "Table 2 - Review Text": table2_review_text,
    "Table 3 - Products": table3_products,
    "Table 4 - Stores": table4_stores,
    "Table 5 - Users": table5_users
}
```

```python
# =========================================================
# 4. SHAPE + COLUMN NAMES
# =========================================================

for name, df in tables.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Shape:", df.shape)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumn Names:")
    for column in df.columns:
        print(" -", column)
```

```python
# =========================================================
# 5. DATA TYPES
# =========================================================

for name, df in tables.items():

    print(f"\n{name}")
    print("-" * 60)

    print(df.dtypes)
```

```python
# =========================================================
# 6. FIRST 5 ROWS
# =========================================================

for name, df in tables.items():

    print(f"\n{name}")
    print("-" * 60)

    display(df.head())
```

```python
# =========================================================
# 7. MEMORY USAGE
# =========================================================

for name, df in tables.items():

    memory_mb = df.memory_usage(deep=True).sum() / (1024 ** 2)

    print(f"{name}: {memory_mb:.2f} MB")
```

```python
# =========================================================
# 8. QUICK SUMMARY TABLE
# =========================================================

summary = pd.DataFrame({
    "Table": list(tables.keys()),
    "Rows": [df.shape[0] for df in tables.values()],
    "Columns": [df.shape[1] for df in tables.values()],
    "Memory (MB)": [
        round(df.memory_usage(deep=True).sum() / (1024 ** 2), 2)
        for df in tables.values()
    ]
})

display(summary)
```
